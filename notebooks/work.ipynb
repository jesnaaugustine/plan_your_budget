```python
doc = fitz.open("/Users/libinvarghese/Documents/Jesna/projects/personal_finacial_assistant/plan_your_budget/data/raw/2dbddc56-6977-4ced-9777-4447dba9b247_invoice.pdf")

page = doc[0]

words = page.get_text('words')

```

```python
rows = {}

for word in words:

    x0, y0, x1, y1, text, *_ = word

    row_key = round(y0)

    if row_key not in rows:
        rows[row_key] = []

    rows[row_key].append((x0, text))
```

```python
for row in rows.values():

    row.sort(key=lambda x: x[0])

    print([text for _, text in row])
```
